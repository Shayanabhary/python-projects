```python
import seaborn as sns
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
```

```python
titanic = sns.load_dataset("titanic")
titanic.to_csv("titanic_dataset.csv", index=False)
df = pd.read_csv("titanic_dataset.csv")
df
```

```python
missing_counts = df.isna().sum()
print(missing_counts)
print("Columns with missing values:", missing_counts[missing_counts > 0].index.tolist())
```

```python
duplicate_mask = df.duplicated()
print(f"duplicated rows count = {duplicate_mask.sum()}")
print("Duplicate row indices:", df.index[duplicate_mask].tolist())
```

```python
numeric_columns = ["survived", "pclass", "age", "fare"]
for column in numeric_columns:
    df[column] = pd.to_numeric(df[column], errors="coerce")
```

```python
survived = df["survived"].value_counts().sort_index()
survival_labels = {0: "not survived", 1: "survived"}
plt.pie(
    survived.values,
    labels=[survival_labels.get(value, str(value)) for value in survived.index],
    autopct="%1.1f%%"
)
plt.title("Titanic survival distribution")
plt.show()
```

```python
pclass = df["pclass"].value_counts().sort_index()
class_labels = {1: "luxury", 2: "medium", 3: "economy"}
plt.bar(pclass.index, pclass.values)
plt.xticks(pclass.index, [class_labels.get(value, str(value)) for value in pclass.index])
plt.xlabel("Passenger class")
plt.ylabel("Number of passengers")
plt.title("Passengers by class")
plt.show()
```

```python
df["pclass"].value_counts().sort_index()
```

```python
gender = df["sex"].value_counts()
gender_labels = {"male": "male", "female": "female"}
plt.pie(
    gender.values,
    labels=[gender_labels.get(value, str(value)) for value in gender.index],
    autopct="%1.1f%%"
)
plt.title("Passenger gender distribution")
plt.show()
```

```python
alone = df["alone"].value_counts().sort_index()
alone_labels = {False: "not alone", True: "alone"}
plt.pie(
    alone.values,
    labels=[alone_labels.get(value, str(value)) for value in alone.index],
    autopct="%1.1f%%"
)
plt.title("Passengers travelling alone")
plt.show()
```
