```python
import seaborn as sns
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
```

```python
titanic = sns.load_dataset("titanic")
titanic.to_csv("titanic_dataset.csv")
df= pd.read_csv("titanic_dataset.csv")
df
```

```python
df.isna().sum()
print(list(df.isna().index))
```

```python
print(f"duplicated rows count = {df.duplicated().sum()}")
print(df[df.duplicated()].index.tolist())
```

```python
survived = df["survived"].value_counts()
plt.pie(survived , labels = ["not survived", "survived"] , autopct = "%1.1f%%" )
plt.show()
```

```python
pclass = df["pclass"].value_counts()
plt.bar(pclass.index, pclass.values)
plt.xticks(pclass.index, ["economy","luxury" ,"medium" ])
plt.show()
```

```python
gender = df["sex"].value_counts()
plt.pie(gender , labels = ["male", "female"] , autopct = "%1.1f%%" )
plt.show()
```

```python
alone = df["alone"].value_counts()
plt.pie(alone , labels = ["alone", "not alone"] , autopct = "%1.1f%%" )
plt.show()
```
